```python
import torch
import matplotlib.pyplot as plt

nomi = ["gatto", "cane", "topo", "leone", "tigre", "elefante"]
emb = torch.tensor([[2., 2.],     # gatto
                    [4., 2.],     # cane
                    [1., 3.],     # topo
                    [9., 9.],     # leone
                    [8., 9.],     # tigre
                    [10., 5.]])   # elefante

plt.figure(figsize=(5, 5))
plt.scatter(emb[:, 0], emb[:, 1])
for nome, (x, y) in zip(nomi, emb):
    plt.text(x + 0.1, y + 0.1, nome)
plt.xlabel("grandezza")
plt.ylabel("selvatichezza")
plt.show()
```

```python
def distanza(i, j):
    return torch.dist(emb[i], emb[j]).item()

print("gatto - cane:  ", distanza(0, 1))
print("gatto - topo:  ", distanza(0, 2))
print("gatto - leone: ", distanza(0, 3))
```

```python
lupo = torch.tensor([5., 10.])   # (grandezza, selvatichezza)

for nome, e in zip(nomi, emb):
    print(nome, torch.dist(lupo, e).item())
```

```python
import torch.nn.functional as F

a = torch.tensor([2., 2.])
b = torch.tensor([4., 4.])
c = torch.tensor([4., 1.])

print("distanza a-b:", torch.dist(a, b).item())
print("distanza a-c:", torch.dist(a, c).item())
print("coseno a-b:", F.cosine_similarity(a, b, dim=0).item())
print("coseno a-c:", F.cosine_similarity(a, c, dim=0).item())
```
