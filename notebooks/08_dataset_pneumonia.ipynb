```python
import numpy as np
import matplotlib.pyplot as plt
from medmnist import PneumoniaMNIST

train_ds = PneumoniaMNIST(split="train", download=True, size=224)
val_ds   = PneumoniaMNIST(split="val",   download=True, size=224)
test_ds  = PneumoniaMNIST(split="test",  download=True, size=224)

print(len(train_ds), len(val_ds), len(test_ds))
print(train_ds.info["label"])

#Tre insiemi, come nella sessione 5:
#Train (per imparare),
#validation (per le decisioni),
#test (da guardare solo alla fine). size=224 chiede le immagini a 224×224 pixel. Il dataset è già diviso da chi l’ha costruito.
```

```python
fig, axes = plt.subplots(2, 4, figsize=(10, 5))
for ax, i in zip(axes.ravel(), range(8)):
    img, label = train_ds[i]
    ax.imshow(img, cmap="gray")
    ax.set_title(f"etichetta = {label[0]}")
    ax.axis("off")
plt.show()
```

```python
for nome, ds in [("train", train_ds), ("val", val_ds), ("test", test_ds)]:
    print(nome, np.bincount(ds.labels.ravel()))
```
